Probability

```python
import numpy as np

# 0 = Tail, 1 = Head

coin_toss = np.random.choice([0,1], size=10)
print(coin_toss)
```

```python
prob_heads = np.mean(coin_toss)
prop_tails = 1 - prob_heads
print(prob_heads, prop_tails)
```

Mean, Median, Variance, Standard Deviation

```python
#data customer purchase amount

data = [2,4,7,10,15]

mean = np.mean(data);
median = np.median(data)
variance = np.var(data)
std_dev = np.std(data)

print(mean, median, variance, std_dev)
```

Normal Distribution

```python
import matplotlib.pyplot as plt
from scipy.stats import norm

mu, sigma = 50, 10

sample = np.random.normal(mu, sigma, 1000)

plt.hist(sample, bins=30, density=True, alpha=0.6, color='g')

x = np.linspace(min(sample), max(sample), 100)

plt.plot(x, norm.pdf(x, mu, sigma),'r')

plt.title("Normal Distribution")
plt.show()
```

Binomial Distribution

```python
from scipy.stats import binom

n,p = 10, 0.5

k = 7

prob_7_heads = binom.pmf(k, n,p)

print("probality of 7 heads in 10 tosses", prob_7_heads)
```

Poisson Distribution

```python
from scipy.stats import poisson

lambda_rate = 3

k = 5

prob_5_cars = poisson.pmf(k, lambda_rate)

print("probability of 5 cars in 1 hour", prob_5_cars)
```

Practice with dataset

```python
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import norm, binom, poisson

np.random.seed(42)

#sample of 100 customer data

data = pd.DataFrame({
    "CustomerId": range(1,101),
    "PurchaseAmount": np.random.normal(50, 10, 100),
    "VisitedWebsite": np.random.choice([0,1],100, p=[0.8, 0.2]),
    "SupportCall": np.random.poisson(3,100)
})

print(data.head())
```

Statistics on dataset

```python
#mean, median, variance, SD for purchase amount

mean_purchase = data['PurchaseAmount'].mean()
median_purchase = data['PurchaseAmount'].median()
variance_purchase = data['PurchaseAmount'].var()
std_dev_purchase = data['PurchaseAmount'].std()

print(f"Mean Purchase Amount: {mean_purchase:.2f}")
print(f"Median Purchase Amount: {median_purchase:.2f}")
print(f"Variance of Purchase Amount: {variance_purchase:.2f}")
print(f"Standard Deviation of Purchase Amount: {std_dev_purchase:2f}")
```

Probality of visiting website

```python
# Probality of visiting website

prob_visited = data['VisitedWebsite'].mean()

print(f"Probability of visiting website: {prob_visited:.2f}")

prob_not_visited = 1 - prob_visited

print(f"Probability of not visiting website: {prob_not_visited:.2f}")
```

```python
plt.hist(data['PurchaseAmount'], bins=15, density=True, alpha=0.6, color='blue')

# Plot normal curve
x = np.linspace(data['PurchaseAmount'].min(), data['PurchaseAmount'].max(), 100)
plt.plot(x, norm.pdf(x, mean_purchase, std_dev_purchase), 'r')
plt.title("Purchase Amount Distribution")
plt.show()
```

```python
from scipy.stats import binom

# Probability of exactly 70 customers visiting
n, p, k = 100, prob_visited, 70
prob_70_visits = binom.pmf(k, n, p)
print(f"Probability exactly 70 customers visited: {prob_70_visits:.4f}")
```

```python
from scipy.stats import poisson

lambda_rate = data['SupportCall'].mean()
k = 3  # probability of exactly 3 calls
prob_3_calls = poisson.pmf(k, lambda_rate)
print(f"Probability a customer made exactly 3 support calls: {prob_3_calls:.4f}")
```
