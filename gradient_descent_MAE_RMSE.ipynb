```python
import numpy as np
import matplotlib.pyplot as plt

#data x is room number, y is rent tk


X = np.array([1, 2, 3, 4, 5])
Y = np.array([1000, 1500, 2000, 2500, 3000])


# init parameter

m, c = 0, 0

learning_rate = 0.01

epochs = 1000

n = len(X)
loss_history = []

# Gradient Descent Lopp

for _ in  range(epochs):

  Y_pred = m * X + c

  error = Y_pred - Y

  # find gradient

  d_m = (2/n) * sum(error * X)
  d_c = (2/n) * sum(error)

  # update parameter

  m -= learning_rate * d_m
  c -= learning_rate * d_c

print(f"Final Model: y = {m:.2f}x + {c:.2f}")
# check prediction

Y_pred = m * X + c


# MAE, RMSE

mae = np.mean(np.abs(Y - Y_pred))
rmse = np.sqrt(np.mean((Y - Y_pred) ** 2))

print("Predictions:", Y_pred)
print("MAE:", mae)
print("RMSE:", rmse)
```
